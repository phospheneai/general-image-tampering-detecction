# Authenta — General Image Forgery Segmentation

Pipeline: **config → pipeline → visualise → train**

## 1 · Setup

```python
import sys
from pathlib import Path

%matplotlib inline

# Project root = parent of this notebook's folder.
PROJ_ROOT = Path(".").resolve().parent
sys.path.insert(0, str(PROJ_ROOT))   # make authgenforge importable in any kernel

CONFIG_PATH = str(PROJ_ROOT / "configs" / "normal" / "train_forensics.yml")

print(f"Config : {CONFIG_PATH}")
```

## 2 · Load pipeline

```python
from authgenforge.options.load import load_pipeline_from_yml

train_loader, test_loader, model, trainer = load_pipeline_from_yml(CONFIG_PATH)

print(f"Experiment : {trainer.experiment_name}")
print(f"Checkpoint : {trainer.ckpt_dir}")
```

## 3 · Visualise batches

Forgery mask overlaid in red — the paired augmentations
(`authgenforge/augmentations/blocks.py`) apply the exact same spatial
transform to the image and mask, so this is the fastest way to catch a
misalignment bug before spending GPU time on a full run.

```python
# --- Train batch ---
from authgenforge.utils.plot_utils import show_mask_batch

train_batch = next(iter(train_loader))
print(f"Train — image: {train_batch['image'].shape}  mask: {train_batch['mask'].shape}")
show_mask_batch(train_batch, label="Train batch", max_images=8)
```

```python
# --- Test batch ---
test_batch = next(iter(test_loader))
print(f"Test  — image: {test_batch['image'].shape}  mask: {test_batch['mask'].shape}")
show_mask_batch(test_batch, label="Test batch", max_images=8)
```

## 4 · Train

```python
trainer.train_model(end_epoch=10)
```
